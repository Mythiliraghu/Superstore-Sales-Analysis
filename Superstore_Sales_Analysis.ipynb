```python
total_sales = df_clean["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_profit = df_clean["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", profit_margin)
```

```python
total_quantity = df_clean["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
total_customers = df_clean["Customer ID"].nunique()

print("Total Unique Customers:", total_customers)
```

```python
total_orders = df_clean["Order ID"].nunique()

print("Total Orders:", total_orders)
```

```python
print("SUPERSTORE BUSINESS SUMMARY")
print("----------------------------")

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
print("Total Quantity:", total_quantity)
print("Unique Customers:", total_customers)
print("Total Orders:", total_orders)
```

```python
category_sales = df_clean.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
category_profit = df_clean.groupby("Category")["Profit"].sum()

print(category_profit)
```

```python
category_analysis = df_clean.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

print(category_analysis)
```

```python
region_sales = df_clean.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
region_profit = df_clean.groupby("Region")["Profit"].sum()

print(region_profit)
```

```python
region_analysis = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

print(region_analysis)
```

```python
segment_analysis = df_clean.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

print(segment_analysis)
```

```python
subcategory_analysis = df_clean.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

print(subcategory_analysis.sort_values(by="Total_Profit", ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Category-wise Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

region_profit.plot(kind="bar")

plt.title("Region-wise Profit Analysis")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales = df_clean.groupby("Year-Month")["Sales"].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(15,6))

plt.plot(monthly_sales.index.astype(str),
         monthly_sales.values,
         marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha=0.5
)

plt.title("Discount vs Profit Analysis")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.axhline(y=0, linestyle="--")

plt.show()
```

```python
plt.figure(figsize=(8,5))

segment_analysis["Total_Sales"].plot(kind="bar")

plt.title("Customer Segment-wise Sales")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
df_clean["Discount_Band"] = pd.cut(
    df_clean["Discount"],
    bins=[-0.001, 0, 0.1, 0.2, 0.3, 0.5, 1],
    labels=["0%", "0-10%", "10-20%", "20-30%", "30-50%", "Above 50%"]
)
```

```python
discount_analysis = df_clean.groupby(
    "Discount_Band", observed=False
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Number_of_Records=("Profit", "count")
)

print(discount_analysis.round(2))
```

```python
highest_sales_month = monthly_sales.idxmax()

highest_month_sales = monthly_sales.max()

print("Highest Sales Month:", highest_sales_month)
print("Sales:", round(highest_month_sales, 2))
```

```python
monthly_seasonality = df_clean.groupby(
    df_clean["Order Date"].dt.month
)["Sales"].sum()

print(monthly_seasonality)
```

# Business Insights

## Finding 1: Technology is the best-performing category

Technology generated the highest sales and profit among the three product categories.

Total Sales: $836,154.03

Total Profit: $145,454.95

Business Recommendation:
The company should continue investing in its Technology category because it contributes significantly to revenue and profit.

## Finding 2: West is the best-performing region

The West region generated the highest sales and profit among all four regions.

Total Sales: $725,457.82

Total Profit: $108,418.45

Business Recommendation:
The company should study the West region's sales strategies and identify opportunities to improve performance in the Central region.

## Finding 3: Consumer segment contributes the most

The Consumer segment generated the highest sales and profit compared to Corporate and Home Office customers.

Total Sales: $1,161,401.34

Total Profit: $134,119.21

Business Recommendation:
The company should continue understanding Consumer customer preferences and develop strategies to retain these customers.

## Finding 4: Some product sub-categories are making losses

The analysis shows that Tables, Bookcases, and Supplies have negative total profits.

- Tables: -$17,725.48
  
- Bookcases: -$3,472.56


- Supplies: -$1,189.10

Business Recommendation: The company should investigate the pricing, production costs, and discount strategies of these products to reduce losses.

## Finding 5: Higher discounts are associated with lower profitability

The analysis indicates that product transactions with discounts above 20% have negative aggregate profit.

Business Recommendation: The company should review its discount policies and evaluate whether high discounts are reducing profitability.

Note: This analysis shows an association, not proof that discounts alone cause losses.

## Finding 6: November records the highest aggregate monthly sales

November generated the highest total sales when comparing the same calendar month across all years in the dataset.

Total November Sales: $352,461.07

Business Recommendation: The company should prepare inventory and marketing campaigns ahead of November to take advantage of the strong sales period.

# Conclusion

The Exploratory Data Analysis of the Superstore Sales dataset helped identify important patterns in sales performance, customer purchasing behaviour, product profitability, and discount strategies.

The analysis revealed that:

1. Technology was the highest-performing product category in terms of sales and profit.
2. The West region generated the highest overall sales and profit.
3. Consumer customers contributed the largest share of sales and profit.
4. Tables, Bookcases, and Supplies experienced negative profitability.
5. Higher discount levels were associated with lower profitability, highlighting the need to review discount strategies.
6. November recorded the highest aggregate sales among the calendar months.

Overall, the analysis demonstrates how data analytics can help businesses understand their performance, identify areas of concern, and make informed decisions.

**Key Takeaway:** Businesses should focus not only on increasing sales but also on improving profitability through effective pricing, inventory management, and customer-focused strategies.
